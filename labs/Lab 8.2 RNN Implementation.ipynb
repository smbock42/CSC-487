```python
class RNN(nn.Module):
    def __init__(self,input_size,hidden_size,output_size):
        super(RNN, self).__init__()
        self.hidden_size = hidden_size

        self.W_ih = nn.Parameter(torch.randn(input_size, hidden_size))
        self.b_ih = nn.Parameter(torch.randn(hidden_size))

        self.W_hh = nn.Parameter(torch.randn(hidden_size, hidden_size))
        self.b_hh = nn.Parameter(torch.randn(hidden_size))

        self.W_ho = nn.Parameter(torch.randn(hidden_size, output_size))
        self.b_ho = nn.Parameter(torch.randn(output_size))

    def forward(self,x):
       # x is a batch of embedding vector sequences of shape [B,N,C] where N is the sequence length, C is the input size
       # forward pass should process the sequence and produce a sequence of N outputs
       # output shape should be [B,N,O] where O is the output size
        batch_size, sequence_length, _ = x.size()
        outputs = torch.zeros(batch_size, sequence_length, self.b_ho.size(0)).to(x.device)
        h_prev = torch.zeros(batch_size, self.hidden_size).to(x.device)

        for t in range(sequence_length):
            x_t = x[:,t,:]
            h_t = torch.tanh(
                x_t @ self.W_ih + self.b_ih +
                h_prev @ self.W_hh + self.b_hh
            )

            y_t = h_t @ self.W_ho + self.b_ho
            outputs[:,t,:] = y_t

            h_prev = h_t

        return outputs


model = RNN(10,20,30)

x = torch.rand(1,5,10)
y = model(x)

y.shape
```
